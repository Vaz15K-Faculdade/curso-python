## Exercício 2
Liste os números e agências de todas as contas bancárias

```python
df_unique = df[['numero_conta_recebedora', 'numero_conta_pagadora']].apply(sorted, axis=1).drop_duplicates()

# Exibir os pares únicos
for conta1, conta2 in df_unique.values:
    print(conta1, conta2)
```

## Exercício 5
Para cada conta, calcule a média (método mean) e o desvio padrão (método std) dos valores transferidos

```python
media = df["valor"].mean()
print(f"Média: {media}")

dp = df["valor"].std()
print(f"Desvio padrão: {dp}")
```

## Exercício 6
Para cada conta, calcule a média (método mean) e o desvio padrão (método std) dos valores recebidos


```python
valores_recebidos = df.groupby('numero_conta_recebedora')['valor'].agg(['mean', 'std'])

print(valores_recebidos)
```

## Exercício 7
Calcule a quantidade e o percentual de ocorrências de cada método de transferência

```python
qtd_ocorrencias = df['método'].value_counts()

print(qtd_ocorrencias)

perc = qtd_ocorrencias / qtd_ocorrencias.sum() * 100

print(perc)
```
