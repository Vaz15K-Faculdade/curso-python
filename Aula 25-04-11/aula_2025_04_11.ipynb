## Exercício 5
Para cada conta, calcule a média (método mean) e o desvio padrão (método std) dos valores transferidos

```python
for conta in tabela["numero_conta_pagadora"].unique():
    print(f"Conta: {conta}")
    print(f"Media: {tabela[tabela['numero_conta_pagadora'] == conta]['valor'].mean()}")
    print(f"Desvio: {tabela[tabela['numero_conta_pagadora'] == conta]['valor'].std()} \n")
```

## Exercício 6
Para cada conta, calcule a média (método mean) e o desvio padrão (método std) dos valores recebidos


```python
for conta in tabela["numero_conta_recebedora"].unique():
    print(f"Conta: {conta}")
    print(f"Media: {tabela[tabela['numero_conta_recebedora'] == conta]['valor'].mean()}")
    print(f"Desvio: {tabela[tabela['numero_conta_recebedora'] == conta]['valor'].std()} \n")
```
