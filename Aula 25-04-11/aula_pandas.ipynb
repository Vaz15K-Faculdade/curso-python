## DataFrames

No Pandas, uma tabela é um DataFrame.

Podemos criar DataFrames a partir de listas ou dicionários, das seguintes formas:

```python
# Criando dataframe a partir de listas
df = pd.DataFrame(
    data=[
        ['Fulano', 20],
        ['Ciclano', 30],
        ['Beltrano', 40]
    ],
    columns=['Nome', 'Idade']
)

df # Para visualizar o dataframe no notebook
```

```python
# Criando dataframe a partir de dicionário de colunas
df = pd.DataFrame(
    data={
        'Nome': ['Fulano', 'Ciclano', 'Beltrano'],
        'Idade': [20, 30, 40]
    }
)

df # Para visualizar o dataframe no notebook
```

```python
# Criando dataframe a partir de lista de dicionários
df = pd.DataFrame(
    data=[
        {'Nome': 'Fulano', 'Idade': 20},
        {'Nome': 'Ciclano', 'Idade': 30},
        {'Nome': 'Beltrano', 'Idade': 40}
    ]
)

df # Para visualizar o dataframe
```

Também podemos importar dados de arquivos CSV e planilhas do Excel. Nesta aula, utilizamos o arquivo ``historico_escolar.csv``. Certifique-se de ter carregado ele nos arquivos da sessão do Colab.

```python
df = pd.read_csv('historico_escolar.csv') # Importação dos dados do arquivo
df # Para visualizar o dataframe
```

Nesta tabela, temos todas as notas trimestrais dos alunos do ensino médio de uma escola. Há uma nota para cada aluno, de cada turma, em cada trimestre e cada disciplina. Há também os índices de falta (índice 0 é 0%, índice 0.5 é 50% e índice 1 é 100%).

Para obter os valores de um coluna específica, ou um grupo de colunas, fazemos o seguinte:

```python
df['nome_aluno'] # Obter apenas a coluna de nomes
```

```python
df[['id_aluno', 'nome_aluno']] # Obter as colunas de nome e id
```

Podemos usar o método Unique para obter o conjunto de valores observados em uma coluna, sem repetições.

```python
df['id_aluno'].unique() # Obter id únicos
```

```python
# Mostrar valores com for
for id_aluno in df['id_aluno'].unique(): # Para cada id único
    print(id_aluno) # Imprimir id
```

Para obter dados de uma única linha a partir do índice dela, usamos iloc

```python
df.iloc[10] # Obtém 11ª linha
```

Para obter uma linha e uma coluna, usamos loc

```python
df.loc[10, 'id_aluno'] # id_aluno da 11ª linha
```

Vamos fazer, no Pandas, o equivalente a uma consulta SQL ``SELECT * FROM historico_escolar WHERE disciplina='História'``, para obter somente as notas de história.

```python
df_historia = df[df['disciplina'] == 'História'] # Consulta SQL
df_historia # Visualizar dataframe
```

Agora vamos obter as notas de história do 1º ano.

Isso aquivale à consulta ``SELECT * FROM historico_escolar WHERE disciplina='História' AND ano_letivo=1``.

```python
df_historia_1ano = df[(df['disciplina'] == 'História') & (df['ano_letivo'] == 1)] # Consulta SQL
df_historia_1ano # Visualizar dataframe
```

Assim como no SQL, no Pandas podemos utilizar agregadores para fazer cálculos como média e mediana.

```python
# Médias
nota_media_geral = df['nota'].mean() # Média de notas
ifalta_medio_geral = df['faltas'].mean() # Média de índices de falta
print(f'Média geral de notas: {nota_media_geral}\nIndice de faltas médio geral: {ifalta_medio_geral*100}%')

# Mediana
nota_mediana_geral = df['nota'].median() # Mediana de notas
faltas_mediana_geral = df['faltas'].median() # Mediana de índices de falta
print(f'Mediana geral de notas: {nota_mediana_geral}\nIndice de faltas médio geral: {faltas_mediana_geral*100}%')
```

Agora, vamos criar uma tabela para visualizar as médias de notas de cada disciplina.

Em SQL, isso seria feito com uma consulta ``SELECT disciplina, AVG(nota) FROM historico_escolar GROUP BY disciplina``.

```python
nota_media_por_disciplina = df.groupby('disciplina')['nota'].mean()
nota_media_por_disciplina # Visualizar dataframe
```

Podemos fazer o group by por vários atributos, e aplicar a agregação a vários atributos também. No seguinte exemplo, criamos uma tabela de médias de notas e índices de falta por disciplina, ano letivo e turma.

```python
tabela_medias = df.groupby(['ano_letivo', 'turma', 'disciplina'])[['nota', 'faltas']].mean()
tabela_medias
```

Nesta tabela criada pelo groupby, os atributos de referência do agrupamento (ano_letivo, turma e disciplina) viram os índices da tabela (ou seja, seus valores são usados para indexar as linhas). Então, para obter uma linha específica dessa tabela, usamos loc especificando o ano letivo, a turma e a disciplina. Veja:

```python
tabela_medias.loc[(2, 1, 'Física')] # 2º ano, turma 1, Física
```

Podemos aplicar expressões aos valores de uma tabela. Nos seguintes exemplos, criamos colunas de aprovação por nota e aprovação por presença.

```python
aprovado_por_nota = df['nota'] >= 5.0 # A nota por trimestre e disciplina deve ser igual ou superior a 5
aprovado_por_nota
```

```python
aprovado_por_freq = df['faltas'] <= 0.5 # Faltas devem ser menos de 50% por trimestre e disciplina
aprovado_por_freq
```

```python
aprovado = aprovado_por_nota & aprovado_por_freq # Aplicando a condição de aprovação
aprovado
```

Podemos inserir essa coluna de reprovação na tabela, da seguinte forma:

```python
df['aprovado'] = aprovado # Criando coluna de aprovação
df
```

Agora, vamos obter apenas os alunos aprovados

```python
df[df['aprovado'] == True]
```
